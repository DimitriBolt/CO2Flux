## 3.4. Действующие критерии допуска I.G.

По [аудиту I.G., стр. 3](../Papers/LEO_CO2_sensor_audit_2026-06-30.pdf#page=3),
допустимый диапазон измерений:

$$0<C_j(t)\leq3000\,\mathrm{ppm}.$$

Служебные значения $C_j(t)\leq-9999\,\mathrm{ppm}$ исключаются;
одиночные выбросы, оставшиеся после диапазонной очистки, удаляются.

Для месячного допуска каждого канала одновременно требуются
([стр. 5](../Papers/LEO_CO2_sensor_audit_2026-06-30.pdf#page=5)):

| Показатель | Критерий I.G. |
|---|---|
| Число допустимых дней за месяц | Не менее 10 |
| Медианная концентрация | 100–2000 ppm |
| Медианная суточная амплитуда | Не менее 8 ppm |

Пороги сохраняются без изменений. Диапазон концентраций не переносится
на амплитуды гармоники. Подгонка глав 1–2 не считается автоматически
методом месячной оценки I.G.; вычислимость фазы не доказывает допуска.
Недостающие определения процедуры перечислены в разделе 3.5.2.
