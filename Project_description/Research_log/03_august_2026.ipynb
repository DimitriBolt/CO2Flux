## 3.1. Какое условие должно предшествовать расчёту фаз

По аудиту I.G. (стр. 3, 5) диапазон допустимых измерений:

$$0<C_j(t)\leq3000\ \mathrm{ppm}.$$

Для месячного допуска одновременно нужны не менее 10 допустимых дней,
медианная концентрация 100–2000 ppm и медианная суточная амплитуда не менее 8 ppm.
Служебные значения и выбросы исключаются; длительные показания выше 3000 ppm
автор относит к плато насыщения.

**Исправление прежнего вывода:** W_R4_C-4 (994/1010/1026) весь август находится
выше 3000 ppm: медианы 4975,80 / 6896,20 / 6971,74 ppm. Поэтому прежние фазы
не допускаю к физической интерпретации. [Расчёт сохранён отдельно](details/archived/august_unqualified_model.ipynb).
Полное временное покрытие не означает пригодности измерений.
