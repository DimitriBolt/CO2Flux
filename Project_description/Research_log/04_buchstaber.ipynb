## 4.1. Метод Бухштабера: программа проверки и реализации

**Суть кандидата.** Представить последовательные фрагменты ряда как точки
многомерной развёртки и проверить, меняется ли их приближение аффинными
подпространствами: ошибка при фиксированной размерности, требуемая
размерность, ориентация подпространства и остаток восстановления.
Это позволяет искать изменения временной организации, которые могут
не проявиться в одной медиане амплитуды или среднем времени максимума.

**Место в маршруте.** Это программа обязательной главы 4 по PLAN.md,
перенесённая из заготовки главы 3 без научного выполнения. Её вход —
пригодные временные ряды LEO, в том числе при наличии исходные концентрации
как самостоятельный ряд. Применение не зависит от положительного результата
главы 3 и не требует надёжных относительных фаз. Сначала воспроизвести
синтетические примеры; затем при выполненных условиях и отдельном разрешении
применить метод к LEO. Положительная дополнительная информативность
не предполагается заранее.

**Геометрическое разграничение для обсуждения с коллегами.**
Три соседние спектральные разности фаз четырёх каналов образуют
при достаточной надёжности точку фазового тора \(\mathbb T^3\)
(циклические углы с отождествлением 0 и \(2\pi\)).
Напротив, оптимальное \(q\)-мерное подпространство
\(L_q^*\) развёртки в \(\mathbb R^n\) принадлежит
грассманиану \(G(n,q)\). Эти пространства, их координаты
и их метрики **не тождественны**. Сопоставление траекторий
относительных фаз с эволюцией оптимальных подпространств
возможно только после отдельного обоснования отображения,
области определённости и сопоставляемых статистик.
Нельзя считать расстояние между подпространствами
эквивалентом изменения относительных фаз и нельзя
приписывать физическую причину ни одной из этих
геометрических характеристик без проверки.

Формулы развёртки, проекции и восстановления приведены по P и B.
Их адаптация к суточным $A_j(d),h_j(d)$, выбор статистических проверок
и критерий обнаружения изменения являются отдельными решениями проекта.
Перед реализацией необходимо сверить используемые теоремы и соглашения
с первоисточниками Бухштабера, перечисленными в разделе 4.3.


**Источник:** [P, §13.1–13.8, стр. 17–20; B, §§11–14, стр. 20–29](../Papers/phase_topology_CO2_LEO.pdf#page=17).

[B: подробная методическая постановка](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=20).

### 4.2.1. Определить вход и два разных масштаба окна

**Предстоящие действия.** Различать длину анализируемого участка $N$
и размерность развёртки $n<N$. Первый задаёт период, которому приписывается
структурная характеристика; второй — длительность памяти внутри него.
Для равномерной сетки $t_k=t_0+(k-1)\Delta t$ она равна $(n-1)\Delta t$.

Рассматривать два самостоятельных варианта входа:
(1) пригодные исходные концентрации $C_j(t)$, чтобы проверять структуру,
которая могла быть потеряна при вычислении суточной гармоники;
(2) готовые суточные характеристики, начиная с отдельных $A_j(d)$.
Пригодность и равномерность календаря проверяются отдельно для каждого входа. Времена $h_j(d)$ нельзя подавать как обычные
линейные числа с разрывом 24→0: возможная адаптация использует две координаты
$u_j(d)=(\cos\theta_j(d),\sin\theta_j(d))$. Такая двумерная запись
представляет одну круговую координату, а не два физических процесса.
Совместная развёртка разных уровней и типов характеристик потребует
явного выбора единиц, масштаба и общей системы координат.

Для векторного ряда $g_k\in\mathbb R^J$ кандидат из P, (51), имеет вид

$$
X_k^{\mathrm{multi}}=(g_k^\mathsf T,\ldots,g_{k+n-1}^\mathsf T)^\mathsf T\in\mathbb R^{Jn}.
$$

Здесь $J$ — число входных координат после выбранного представления;
совместный вариант проверяется после отдельных рядов.

При дневном входе $\Delta t=1$ сутки; выбор $n$ в днях или неделях не является
сменой периода 24 ч у исходной гармоники. Значение $n=97$ для 15-минутных
данных из P нельзя переносить на дневные оценки. Ни $N$, ни $n$ здесь
численно не назначаются.

**Проверка перед построением.** Отсутствующий день разрывает регулярный ряд:
нельзя сжать календарь, удалить пропуски и объявить оставшиеся 783 точки
равномерными. Использовать непрерывные допустимые фрагменты и строить
только полностью наблюдённые лаговые окна. Если их мало, зафиксировать
невозможность выбранного масштаба. Окна двух разделённых фрагментов не соединять.
Масштабирование должно быть одинаковым в сравниваемых годах; отдельная
нормировка каждого окна может скрыть изменение амплитуды.

**Результат шага.** Паспорт входа: $f$, единицы, сетка, маска, $N$, $n$,
правило масштабирования и число доступных окон. Развёртка исходных
внутрисуточных концентраций $C_j(t)$ — другой вход, необходимый для будущей
проверки сохранения самой суточной гармоники после реконструкции.


**Источник:** [P, §13.2, стр. 17, (40)–(42); §13.6, стр. 19, (51); B, §14, этап A, стр. 27](../Papers/phase_topology_CO2_LEO.pdf#page=17).

Адаптация к суточным характеристикам и календарным пропускам сформулирована для этой главы.

### 4.2.2. Построить развёртку и проверить центрирование

Для одного непрерывного скалярного фрагмента:

$$
X_k=(f_k,\ldots,f_{k+n-1})^\mathsf T\in\mathbb R^n,
\quad k=1,\ldots,P,\quad P=N-n+1,
\qquad \mathcal H=[X_1\ \cdots\ X_P].
$$

Последовательность $X_k$ задаёт ориентированную ломаную развёртки.
Соседние точки имеют $n-1$ общих наблюдений и статистически зависимы.
Здесь $\mathcal H$ — матрица развёртки; она не совпадает с вектором
времён максимумов $\mathbf H_v(d)$.

$$
\bar X=\frac1P\sum_{k=1}^{P}X_k,\qquad
\mathcal H_c=\mathcal H-\bar X\mathbf1_P^\mathsf T,
\qquad S=\frac1P\mathcal H_c\mathcal H_c^\mathsf T.
$$

**Что проверить.** Центрируется вектор лаговых координат $\bar X$;
вычитание одного среднего всего скалярного ряда вообще не эквивалентно
этой операции на конечном участке. Аффинный ранг относится к
$\mathcal H_c$, исходный ранг Hankel-матрицы — к $\mathcal H$.
Нормировка $1/P$ здесь выбрана явно. Для объединения нескольких фрагментов
формула $P=N-n+1$ не применяется к суммарному числу точек: считаются только
допустимые окна внутри каждого фрагмента.

**Результат шага.** Проверенные лаговые индексы, $\bar X$, $\mathcal H_c$
и $S$. Сохранять $\bar X$ отдельно: центрирование не должно скрыть изменение
уровня ряда, которое оценивается вместе с изменением структуры.


**Источник:** [P, §13.2–13.3, стр. 17–18, (41)–(43); B, §11, стр. 20, (61)–(63), §12, стр. 21, (68)–(69)](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=20).

### 4.2.3. Вычислять весь профиль ошибок и различать виды ширины

Пусть $\mathcal H_c=U\Sigma V^\mathsf T$,
$\lambda_i=\sigma_i^2/P$, $\lambda_1\geq\cdots\geq\lambda_n\geq0$.
Если сингулярных чисел меньше $n$, спектр дополняется нулями.
Для принятой нормировки средняя квадратичная ошибка оптимальной
аффинной проекции размерности $q$ равна

$$
E_q=\min_{\dim L=q}\frac1P\sum_{k=1}^{P}
\|X_k-\bar X-\pi_L(X_k-\bar X)\|_2^2
=\sum_{i=q+1}^{n}\lambda_i.
$$

$$
W_q^{(2)}=P E_q,\qquad d_{q,\mathrm{RMS}}=\sqrt{E_q},
\qquad \eta_q=\frac{E_q}{\mathrm{tr}\,S}\quad(\mathrm{tr}\,S>0).
$$

$W_q^{(2)}$ — суммарная квадратичная ошибка, $E_q$ — средняя квадратичная,
$d_{q,\mathrm{RMS}}$ имеет единицы ряда; $\eta_q$ безразмерна.
В P используется нормированная матрица, в B, (69)–(71), ненормированная:
их численные значения нельзя сравнивать без множителя $P$.
Для амплитуд единицы $E_q$ — ppm². Сохранять весь профиль $\eta_q$,
а не только размерность при одном пороге.
При доле объяснённой дисперсии $V_q=\sum_{i=1}^q\lambda_i/\mathrm{tr}\,S$
имеем $\eta_q=1-V_q$: это та же спектральная информация PCA, а не новый
независимый признак. Дополнительные вопросы связаны с ориентацией,
восстановлением и изменением этих величин во времени.

Максимальное отклонение от аффинного подпространства задаёт другую задачу:

$$
d_q^{\max}=\inf_{a,\,\dim L=q}\max_k
\mathrm{dist}(X_k,a+L).
$$

Хвостовая сумма собственных значений решает квадратичную задачу и не является
автоматически решением этой задачи на максимум. Её отдельное вычисление —
возможное последующее расширение, требующее собственной проверки оптимизации.

Дополнительные сводки, если они понадобятся:

$$
q_\varepsilon=\min\{q:\eta_q\leq\varepsilon\},\qquad
p_i=\frac{\lambda_i}{\mathrm{tr}\,S},\qquad
r_{\mathrm{ent}}=\exp\!\left(-\sum_i p_i\log p_i\right),
$$

где $0\log0=0$. Порог $\varepsilon$ следует обосновать по шуму и проверочным
данным до интерпретации. При нулевой дисперсии относительные показатели
не определены; постоянный ряд обрабатывается отдельно.
Относительный критерий $\eta_q\leq\varepsilon$ не тождествен абсолютному
пороговому критерию для ненормированной хвостовой суммы.

**Результат шага.** Спектр, суммарная дисперсия, профили $E_q$, $\eta_q$
и обоснованные сводки. Они характеризуют эмпирическое аффинное приближение,
не число механизмов и не порядок физической модели.


**Источник:** [P, §13.3, стр. 18, (43)–(46); B, §§11–13, стр. 20–23, (66)–(77)](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=21).

### 4.2.4. Сравнивать подпространства, а не знаки отдельных векторов

Оптимальное для $E_q$ подпространство и его проектор:

$$
L_q^*=\mathrm{span}(u_1,\ldots,u_q)\in G(n,q),
\qquad \Pi_q=U_qU_q^\mathsf T.
$$

Для двух окон с одинаковыми координатами, масштабом и размерностью $q$
главные углы определяются через сингулярные числа:

$$
\cos\vartheta_i=\sigma_i(U_{q,1}^\mathsf T U_{q,2}),
\qquad d_G(L_{q,1}^*,L_{q,2}^*)=
\left(\sum_{i=1}^{q}\vartheta_i^2\right)^{1/2}.
$$

**Что проверить.** Знаки и базисы в одном подпространстве произвольны;
их смена не должна считаться изменением данных. При $q<n$ единственность
выбранного подпространства требует $\lambda_q>\lambda_{q+1}$;
малый разрыв между этими значениями делает оценку чувствительной к шуму.
Сохранять этот разрыв и bootstrap-неопределённость проектора.
Нельзя сравнивать $d_G$ для разных $n$, разных наборов координат или
разных правил нормировки. При разном $q$ необходим отдельный способ сравнения.

**Результат шага.** Расстояния между сопоставимыми окнами и их неопределённость.
Равенство спектров не означает равенства подпространств; их поворот
не устанавливает причину изменения.


**Источник:** [P, §13.3, стр. 18, (44), и §13.7, стр. 20, (52); B, §13, стр. 25, (87)–(88)](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=25).

### 4.2.5. Восстановить ряд через L-приближение

После выбора $L$ спроецировать каждый узел:

$$
Y_k(L)=\bar X+\pi_L(X_k-\bar X).
$$

Перекрывающиеся оценки одной исходной точки затем согласуются усреднением
по антидиагоналям. Если $Y_{i,k}$ — координата $i$ узла $k$, то

$$
I_s=\{(i,k):i+k-1=s\},\qquad
\widehat f_s(L)=\frac1{|I_s|}\sum_{(i,k)\in I_s}Y_{i,k}(L),
\qquad r_s(L)=f_s-\widehat f_s(L).
$$

**Что проверить.** На краях число слагаемых отличается от центральной части.
Проекция на все доступные направления должна воспроизводить исходный ряд
с точностью численного округления. Согласование перекрытий проводится
отдельно внутри каждого непрерывного фрагмента; разрывы остаются разрывами.
После антидиагонального усреднения нельзя автоматически приписывать
восстановленному ряду исходное ограничение ранга.

Минимизация ошибки проекции лаговых узлов и минимизация ошибки
восстановленного ряда — разные задачи. PCA решает первую; оптимальность
для второй из этого не следует. Для фазового входа в координатах
$(\cos\theta,\sin\theta)$ реконструкция может выйти с единичной окружности:
длину восстановленного вектора и влияние обратного отображения нужно
проверить отдельно; при нулевом векторе угол не определён.

**Результат шага.** Исходный ряд, восстановление, остаток и прослеживаемые
индексы. Остаток исследуется как неописанная часть выбранного приближения,
без автоматического отнесения к биологическому или химическому источнику.


**Источник:** [P, §13.4–13.5, стр. 18–19, (47)–(50); B, §13, стр. 26–27, (90)–(93)](../Papers/phase_topology_CO2_LEO.pdf#page=19).

### 4.2.6. Проверять изменение структуры на сопоставимых окнах

**Предстоящие действия.** Использовать одинаковые $n$, единицы, нормировку
и правила отбора для сравнения календарно сопоставимых участков.
В каждом окне сохранять $\bar X$, $\mathrm{tr}\,S$, полный профиль $\eta_q$,
$L_q^*$ и остаток восстановления. Возможные описательные различия:

$$
\Delta\eta_q=\eta_q(W_2)-\eta_q(W_1),\qquad
\Delta E_q=E_q(W_2)-E_q(W_1),\qquad d_G(L_q^*(W_1),L_q^*(W_2)).
$$

Различать три ситуации из B: меняется общий масштаб вариации при близком
$\eta_q$; меняется доля ошибки при фиксированном $q$; меняется ориентация
при близких спектрах. Слово «близкий» требует заранее установленной
оценки неопределённости и не определяется впечатлением от графика.

Дополнительный кандидат — выучить опорное подпространство $L_{\mathrm{ref}}$
на одном периоде и проверить ошибку на другом без подстройки:

$$
E_{\mathrm{ref}}(W_2)=\frac1{P_2}\sum_k
\|X_{2,k}-\bar X_2-\pi_{L_{\mathrm{ref}}}(X_{2,k}-\bar X_2)\|_2^2.
$$

Эта версия сравнивает центрированную структуру; изменение $\bar X_2-\bar X_1$
регистрируется отдельно. Если проверяется перенос всей аффинной модели
с фиксированным центром $\bar X_1$, её ошибку нужно обозначить отдельно.
Такой перенос — адаптация предложенного в B опорного подпространства.

**Будущий результат.** Свидетельство устойчивого изменения выбранного
эмпирического представления либо отсутствие такого свидетельства.
Сравнить пользу метода с [простыми статистическими кандидатами главы 3](03_long_term_evolution.ipynb#candidate-methods).
Изменение ширины или ориентации не является само по себе доказательством
почвообразования; влияние условий наблюдения остаётся ограничением.


**Источник:** [B, §13, стр. 22–26, (79)–(88), опорное подпространство на стр. 26; P, §13.7–13.8, стр. 19–20](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=22).

### 4.2.7. Проверить реализацию на синтетических примерах

До LEO необходимо воспроизвести случаи с известной геометрией.
Следующая таблица — программа будущих проверок, а не отчёт о пройденных тестах.

| Пример или преобразование | Что необходимо проверить |
|---|---|
| Постоянный ряд | После центрирования аффинный ранг 0; исходная ненулевая Hankel-матрица может иметь ранг 1; относительная ширина не определена |
| Линейный тренд | Аффинный ранг не выше 1 при корректном центрировании; проверка непосредственно из определения развёртки |
| Одна синусоида | Аффинный ранг не выше 2; при невырожденной сетке и достаточных $n,P$ ожидается 2; вырожденная выборка может понизить ранг |
| Сдвиг и масштаб $g_k=a f_k+b$, $a\ne0$ | $\mathcal H_c(g)=a\mathcal H_c(f)$, собственные значения масштабируются на $a^2$, нормированные профили сохраняются при ненулевой дисперсии; проектор сохраняется при единственности выбранного подпространства |
| Полная проекция и перекрытия | Восстановление всех точек, включая края; корректные количества слагаемых на антидиагоналях |
| Знаки базиса и близкие собственные значения | Смена знака не меняет проектор; неустойчивость при малом спектральном разрыве отражается в неопределённости |
| Шум, пропуски, искусственный сдвиг режима | Не появляются окна через разрывы; оцениваются ложные обнаружения и чувствительность к известному изменению |

Для $f_k=c+a\cos(\omega k\Delta t+\varphi)$ низкий ранг — следствие
конкретной структуры сигнала. Обратное утверждение неверно: ранг 2
сам по себе не доказывает синусоидальность, периодичность или диффузию.
Точности численных сравнений и масштабы шума предстоит задать до тестирования.

**Результат шага.** Воспроизводимые синтетические примеры и перечень условий,
при которых реализация корректна или неприменима. Сначала проверяется
скалярный случай, затем многоканальная адаптация с фиксированным масштабированием.


**Источник:** [B, §13, стр. 27, (94); §29, пункты 3–5, стр. 47; P, §13.3, стр. 18](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=27).

Проверки тренда, сдвига, масштаба и индексов здесь явно сформулированы из определений для будущей реализации.

### 4.2.8. Проверить неопределённость и сохранение амплитуд и фаз

**Предстоящие действия.** Повторить оценку при заранее выбранных $N,n,q$,
оценить устойчивость профиля ширин и проекторов блочным bootstrap.
Перекрывающиеся узлы не считать независимыми наблюдениями. При пересборке
временных блоков не создавать лаговые окна, пересекающие искусственные
стыки несмежных блоков. Проверить перенос на отложенный сопоставимый период
без изменения настроек. Для совместного входа сравнить с отдельными
развёртками уровней; исключение датчика меняет пространство и требует
сопоставления определённых заранее выходных показателей.

На синтетических рядах, а позднее на допущенных исходных концентрациях,
проверять сохранение характеристик одной и той же гармонической процедурой:

$$
e_{A,j}(d)=\widehat A_j(d)-A_j(d),\qquad
e_{h,j}(d)=\mathrm{wrap}_{24}\bigl(\widehat h_j(d)-h_j(d)\bigr).
$$

Малая ошибка восстановления концентраций не гарантирует малой ошибки фазы.
При неопределённой исходной или восстановленной фазе $e_{h,j}$ не вычисляется
как обычное число. Если входом были дневные $A_j$ и круговые координаты $h_j$,
проверяется сохранение именно этих входных характеристик; это не заменяет
проверки реконструкции внутрисуточного сигнала.

В источниках также предложены суррогатные ряды с сохранённым спектром
и разрушенными фазовыми связями. Это возможный дополнительный контроль
против заранее заданной нулевой гипотезы, а не универсальное доказательство
дрейфа. Генератор, сохранённые свойства, допустимая регулярная сетка
и критерий сравнения должны быть описаны до применения.

**Результат шага.** Неопределённость структурных характеристик, ошибки
сохранения амплитуд и фаз, область устойчивых настроек и результаты
отложенной проверки. Прохождение проверки амплитуд не подменяет проверку фаз.
Приведённая выше ошибка \(e_{h,j}\) относится к **исторической**
24-часовой характеристике времени максимума, а не к новой целевой
спектральной разности фаз. Когда I. R. Gabitov явно согласует
рабочую частоту, если будет отдельно утверждена такая проверка,
оценивать сохранение именно \(\Delta\phi_{0,5}\),
\(\Delta\phi_{5,20}\), \(\Delta\phi_{20,35}\)
по одинаковому спектральному протоколу на исходном и
восстановленном сигнале. До его решения не объявлять
проверку прежней \(e_{h,j}\) проверкой новых относительных фаз.


**Источник:** [P, §13.5–13.8, стр. 19–20, §20.2, стр. 26; B, §29, пункты 4–11, стр. 47](../Papers/phase_topology_CO2_LEO.pdf#page=19).

### 4.2.9. Отдельные кандидаты после проверки базовой реализации

**Локальное продолжение ряда.** B предлагает оценивать аффинную рекуррентную
связь на стабильном окне и проверять короткий прогноз за его пределами:

$$
\widehat f_t=\beta_0+\sum_{\ell=1}^{p}\beta_\ell f_{t-\ell},\qquad
E_{\mathrm{pred}}=\frac1K\sum_{t\in W_{\mathrm{test}}}
(f_t-\widehat f_t)^2.
$$

Для многошагового прогноза необходимо заранее указать использование
предыдущих прогнозов или наблюдений. Резкий рост ошибки — кандидат
на нарушение эмпирической регулярности. У шумного ряда это эвристика,
а не применение точной теоремы о сохранении ранга без проверки её условий.
Порядок $p$, регуляризация и горизонт выбираются на обучающем участке;
проверочный участок не используется для подбора.

**Выбор подпространства с контролем фаз.** P, (47), и B, (83), предлагают
проектное расширение критерия проекции. Возможная безразмерная запись:

$$
J(L)=E_{\mathrm{rec}}(L)+\alpha E_{\mathrm{phase}}(L),
\qquad \alpha\geq0,
$$

где $E_{\mathrm{rec}}=\sum_s(f_s-\widehat f_s)^2/\sum_s(f_s-\bar f)^2$
определено при ненулевом знаменателе, а $E_{\mathrm{phase}}$ — заранее
описанная средняя квадратичная круговая ошибка на общих допустимых датах,
например среднее $(e_{h,j}/(24\,\mathrm{ч}))^2$. Для нескольких каналов
необходимо явно задать их масштабирование и суммирование ошибок.
В PDF фазовый штраф сформулирован для выбранных фазовых отношений;
использование ошибки времени максимума здесь является отдельной адаптацией.

Нужно сравнить PCA-базу, оптимизацию ошибки реконструкции при $\alpha=0$
и фазовый штраф при $\alpha>0$ на отложенных данных. Первые два варианта
вообще не обязаны совпадать. $\alpha$ не выбирается по желаемому научному
выводу; оптимум численного алгоритма требует собственной проверки.
Это расширение проекта LEO, а не теорема или готовый алгоритм Бухштабера.
Физические штрафы, $D$, $\kappa$ и геометрическая фаза требуют последующих
этапов и здесь не вводятся в вычисление.

**Будущий результат.** Отдельное обоснование, даёт ли каждое расширение
дополнительную информацию сверх базовой развёртки, с указанием неудачных случаев.


**Источник:** [B, §13, стр. 24–26, (83), (89), §14, этап C, стр. 28, (97); P, §13.4, стр. 18, (47)](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=24).

### 4.2.10. Что должно остаться после тщательной проверки метода

1. Сверенные определения и первоисточники; паспорт входа, сетки, масок,
   масштабирования, $N$, $n$ и нормировки матрицы.
2. Синтетические проверки центрирования, ранга, ширины, проекторов
   и восстановления, включая вырожденные случаи и пропуски.
3. Спектры, полные профили ошибок, спектральные разрывы, расстояния
   между сопоставимыми подпространствами и их неопределённость.
4. Реконструкции, остатки, отдельные проверки сохранения амплитуд и фаз,
   проверка переноса на отложенный период.
5. Сопоставление с простыми статистическими кандидатами и заключение
   о дополнительной информативности либо установленном ограничении метода.

Каждое утверждение о структурном изменении должно иметь конкретные даты,
характеристику изменения, оценку неопределённости и проверку чувствительности.
Выбор шага по графику, увеличение ранга или одно большое $d_G$ сами по себе
не являются научным заключением. Отрицательный результат сохраняется;
он не служит основанием объявить применение обязательного метода выполненным
до прохождения согласованной программы проверки.


**Источник:** [P, §13.8, стр. 20, §20, стр. 26; B, §14, стр. 27–29, §29, стр. 47](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=27).
