Для такой проверки прежде нужно выяснить, имеются ли вообще воспроизводимые длительные изменения характеристик CO₂, которые можно обоснованно обсуждать.

## 3.4. Математические объекты и неизменяемые соглашения

Сохраняются определения и метод [главы 2](02_monthly_changes.ipynb) для фиксированной колонки $v=\mathrm{W\_R4\_C{-}4}$:

$$
\mathbf H_v(d)=\bigl(h_{v,1}(d),h_{v,2}(d),h_{v,3}(d)\bigr)\in\mathbb T_{24}^3,
\qquad
\mathbf A_v(d)=\bigl(A_{v,1}(d),A_{v,2}(d),A_{v,3}(d)\bigr)\in[0,\infty)^3\,\mathrm{ppm}.
$$

Здесь $h_{v,j}(d)$ — **время максимума** выделенной суточной гармоники, а $A_{v,j}(d)$ — её амплитуда, не концентрация и не максимум $c+A$. Период исходной гармоники **24 часа фиксирован**; пересмотр периода не входит в задачу. Времена максимумов — циклические координаты: $0$ и $24$ часа совпадают. Поэтому обычная линейная интерпретация их перехода через полночь некорректна.

Координаты точек берутся из готовых результатов главы 2. Публикуемые основания месячного допуска, собственная вычислительная пригодность суток и физическая надёжность отдельной фазы **не отождествляются**. Не вводить новый автоматический порог $R^2$, новые фильтры или заполнение пропусков без согласования в рамках PLAN.md.

Траектории $\mathbf H_v(d)$ и $\mathbf A_v(d)$ — разные пространства и результаты. Анализ самостоятельной геометрии **двух относительных фаз** предусмотрен главой 5 при наличии надёжных фазовых характеристик, а не текущей главой. #### 3.5.5.1. Парное календарное сопоставление

Пусть $K_{m;y_1,y_2}$ — числа месяца $m$, присутствующие в основном допущенном массиве обоих лет, $y_2>y_1$. Все пары лет задаются календарными границами основного блока, включая сравнения с недостаточным покрытием. Для $k\in K$:

$$
d_{A,j}(k)=A_j(y_2,m,k)-A_j(y_1,m,k),\qquad
\Delta A_j=\mathrm{median}_{k\in K}d_{A,j}(k).
$$

Сравнение допускается при $|K|\geq10$ и наличии минимум двух ISO-недель (понедельник–воскресенье) **в каждом году**. Сохраняются оба списка дат и причины отсутствия остальных чисел. Сентябрь 2013, три пограничные даты и 1244 предварительные точки не входят в $K$. Не меняются границы месяца и не заполняются пропуски.

**Источники:** [B, §5, шаг I.11, стр. 15](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=15), [§29, п. 13, стр. 47](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=47); [P, §19, стр. 25](../Papers/phase_topology_CO2_LEO.pdf#page=25). Парные даты и медиана разностей — утверждённая адаптация; одинаковый месяц не гарантирует одинакового гидрологического состояния.

#### 3.5.5.2. Парные круговые разности времени максимума

$$
\delta_j(k)=\mathrm{wrap}_{24}\bigl(h_j(y_2,m,k)-h_j(y_1,m,k)\bigr),\qquad
\mathrm{wrap}_{24}(u)=((u+12)\bmod24)-12,
$$

$$
z_j=\frac1{|K|}\sum_{k\in K}e^{2\pi i\delta_j(k)/24},\qquad
R_j=|z_j|,\qquad
\Delta h_j=\frac{24}{2\pi}\arg z_j\in[-12,12).
$$

В этих формулах время выражено в часах. Все парные сутки имеют одинаковый вес. Сохраняются круговые разности, среднее и концентрация $R_j$; знак исходной гармонической фазы не переопределяется. При численно нулевом результанте ($R\leq10^{-12}$, только допуск арифметики) угол не определён. Слабая концентрация и разброс bootstrap-средних проверяются совместно; новый эмпирический порог исправности датчика или $R^2$ не вводится.

Круговой интервал строится вокруг оценённого направления, а не по линейно отсортированным часам через полночь (3.5.5.3). При отсутствии локально определённого интервала положительный фазовый вывод запрещён. Существующие отметки надёжности каждой исходной фазы сохраняются; статистическая устойчивость среднего не доказывает физической надёжности подгонки.

**Источник:** [P, §20.2, стр. 26, формула (66)](../Papers/phase_topology_CO2_LEO.pdf#page=26). Применение кругового среднего непосредственно к парным разностям утверждено для главы 3.

#### 3.5.5.3. Блочный bootstrap и круговая неопределённость

Параметры: $B=2000$, основной блок $\ell=7$ суток, чувствительность $\ell=3,14$, уровень 95%, фиксированное начальное зерно **20260929**. Единица пересэмплирования — вектор парных разностей трёх датчиков, а не независимые годы или датчики.

На упорядоченном $K$ формируются все перекрывающиеся блоки из $\ell$ **последовательных календарных суток**, целиком лежащие внутри одного непрерывного фрагмента в обоих годах. Блоки выбираются равновероятно с возвращением, соединяются до длины $|K|$, последний обрезается до этой длины. Статистики используют полученные значения; искусственные стыки не трактуются как смежные наблюдения. Для трёх датчиков и двух характеристик используются одинаковые индексы выборки.

Чтобы не скрыть исключение коротких фрагментов, интервал для всего $K$ рассчитывается только если **каждая парная дата представлена хотя бы в одном полном блоке** и доступны минимум два различных блока. Иначе сохраняются точечные оценки, число/длины фрагментов и причина недоступности данного $\ell$. Короткие фрагменты не отбрасываются ради интервала; основной блок 7 не заменяется автоматически на 3. Это условие применимости выбранного bootstrap, не новый допуск датчиков.

$$
I_A=[Q_{0.025}(\Delta A^*),Q_{0.975}(\Delta A^*)].
$$

Для фазы вычисляются $e_j^*=\mathrm{wrap}_{24}(\Delta h_j^*-\Delta h_j)$ и центральный процентильный интервал отклонений $[q_{.025},q_{.975}]$. Он переносится на окружность как дуга

$$
I_h=\Delta h_j+[q_{.025}(e_j^*),q_{.975}(e_j^*)]\pmod{24}.
$$

Квантили — линейная интерполяция порядковых статистик. Локальная дуга принимается только если все 2000 средних определены и центральный 95%-й интервал отклонений лежит строго внутри $(-6,6)$ ч — открытой полуокружности вокруг оценки. Иначе фиксируется неустойчивость/неопределённость кругового интервала без положительного фазового вывода. Предел 6 ч выражает геометрию полуокружности, а не критерий исправности датчика. Для проверки наличия нуля используется вся дуга, включая переход через границу $\pm12$ ч. Вместе с дугой сохраняются $R$, концентрация bootstrap-направлений и число неопределённых повторов.

**Источники:** [P, §20.1–20.2, стр. 26](../Papers/phase_topology_CO2_LEO.pdf#page=26); [B, §29, п. 6, стр. 47](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=47). Указанные параметры и реализация на неполном парном календаре конкретизируют утверждённую процедуру.

#### 3.5.5.4. Воспроизводимость и чувствительность к покрытию

Для каждого месяца сравниваются результаты всех доступных пар лет при $\ell=7$ и допустимых $\ell=3,14$. Указываются направление, величина, интервалы и их доступность. Сравнения, имеющие общий год, помечаются как **зависимые**; они не увеличивают число независимых подтверждений. Отсутствие сравнимой другой пары является ограничением воспроизводимости.

Для проверки покрытия определяется $K_m^{\mathrm{all}}=\bigcap_y K_{y,m}$ — общий набор чисел месяца **всех лет этого месяца внутри основного блока**, включая год с отсутствующими допустимыми датами. На нём повторяются те же парные статистики и bootstrap, только если выполнены прежние условия 10 суток/две недели и непрерывности. Это заранее заданная проверка покрытия, не поиск более выгодного интервала. Сохраняются изменение числа пар и сдвиг оценки относительно основного $K$:

$$
S_{A,j}=\Delta A_j(K_m^{\mathrm{all}})-\Delta A_j(K),\qquad
S_{h,j}=\mathrm{wrap}_{24}\bigl(\Delta h_j(K_m^{\mathrm{all}})-\Delta h_j(K)\bigr).
$$

В выводе отдельно перечисляются воспроизводимые особенности амплитуд и времён максимумов, зависимость от календарного покрытия и невозможные проверки. Ни отдельный межгодовой контраст, ни отсутствие устойчивого контраста не доказывают наличие или отсутствие необратимых физических изменений. Дополнительные фильтры, гармонические расчёты, суррогаты и другие методы не вводятся.

**Источники:** [B, §29, стр. 47](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=47); [P, §5.2, стр. 8](../Papers/phase_topology_CO2_LEO.pdf#page=8), [§20, стр. 26](../Papers/phase_topology_CO2_LEO.pdf#page=26). Конкретная проверка общего календарного покрытия задана здесь до вычислений.
