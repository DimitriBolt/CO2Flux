## 3.1. Введение и место в исследовании

Конечная задача исследования — проверить возможность различения вклада переноса CO₂, химического взаимодействия с базальтом и биологических источников и стоков по эволюции наблюдений. Для такой проверки прежде нужно выяснить, имеются ли вообще воспроизводимые длительные изменения характеристик CO₂, которые можно обоснованно обсуждать.

В главе 1 получен исходный пример суточной гармонической модели. Глава 2 обобщила вычисление на выбираемые колонку и временной интервал; для исторической вертикали **W_R4_C-4** получены суточные векторы времён максимумов $\mathbf H_v(d)$ и амплитуд $\mathbf A_v(d)$ и представлены две трёхмерные траектории с сохранением пропусков и статусов допуска.

Однако построение траекторий и визуальное различие отдельных участков ещё не доказывают длительной эволюции: наблюдаемая картина может зависеть от отсутствующих суток, календарного покрытия, цикличности времён максимумов и надёжности подгонки. Настоящая глава должна перейти от описания готовых траекторий к обоснованной проверке их долговременных изменений.

В соответствии с [PLAN.md](PLAN.md#trajectories) анализ выполняется **на уже полученных исторических результатах**. Новые актуальные датчики, август 2026 года и допуск W_R4_C1 не являются необходимыми входами этого этапа.

## 3.2. Цель и критерий завершения

**Цель:** установить, позволяют ли уже рассчитанные суточные характеристики исторической вертикали W_R4_C-4 обнаружить **воспроизводимые длительные изменения амплитуд и времён максимумов** с учётом опубликованного основания допуска I.G., пропусков и ограничений надёжности фаз.

Вопросы главы:
1. Какие изменения характеристик наблюдаются на основном историческом блоке, если сохранять реальное календарное расположение пригодных суток?
2. Остаются ли наблюдаемые изменения обоснованными после учёта цикличности времён максимумов, разрывов и известных ограничений оценки?
3. Достаточно ли имеющихся данных для утверждения о воспроизводимой длительной эволюции или допустимо только констатировать её неустановленность?

**Критерий завершения, без дополнений к PLAN.md:** получены воспроизводимые длительные изменения амплитуд и фаз **либо** обоснованное заключение, что имеющиеся данные не позволяют их установить.

Определение конкретной процедуры проверки воспроизводимости должно быть явно документировано **до новых расчётов**. Разделы 3.5.5–3.5.6 описывают кандидатов на проверку и реализацию; конкретные пороги и окончательная процедура ещё не выбраны. После главы 3 обязательным отдельным этапом является метод Бухштабера (глава 4):
его применение к пригодным временным рядам LEO не зависит от положительного
результата по суточным амплитудам или надёжности относительных фаз.
Самостоятельная геометрия относительных фаз относится к главе 5 при
наличии надёжного фазового входа. Физическая атрибуция остаётся главой 6.

## 3.4. Математические объекты и неизменяемые соглашения

Сохраняются определения и метод [главы 2](02_monthly_changes.ipynb) для фиксированной колонки $v=\mathrm{W\_R4\_C{-}4}$:

$$
\mathbf H_v(d)=\bigl(h_{v,1}(d),h_{v,2}(d),h_{v,3}(d)\bigr)\in\mathbb T_{24}^3,
\qquad
\mathbf A_v(d)=\bigl(A_{v,1}(d),A_{v,2}(d),A_{v,3}(d)\bigr)\in[0,\infty)^3\,\mathrm{ppm}.
$$

Здесь $h_{v,j}(d)$ — **время максимума** выделенной суточной гармоники, а $A_{v,j}(d)$ — её амплитуда, не концентрация и не максимум $c+A$. Период исходной гармоники **24 часа фиксирован**; пересмотр периода не входит в задачу. Времена максимумов — циклические координаты: $0$ и $24$ часа совпадают. Поэтому обычная линейная интерпретация их перехода через полночь некорректна.

Координаты точек берутся из готовых результатов главы 2. Публикуемые основания месячного допуска, собственная вычислительная пригодность суток и физическая надёжность отдельной фазы **не отождествляются**. Не вводить новый автоматический порог $R^2$, новые фильтры или заполнение пропусков без согласования в рамках PLAN.md.

Траектории $\mathbf H_v(d)$ и $\mathbf A_v(d)$ — разные пространства и результаты. Анализ самостоятельной геометрии **двух относительных фаз** предусмотрен главой 5 при наличии надёжных фазовых характеристик, а не текущей главой. Между ними обязательная самостоятельная глава 4 посвящена методу Бухштабера.

#### 3.5.5.1. Межгодовое сравнение календарно сопоставимых участков

**Суть.** Проверить, сохраняются ли различия характеристик при сопоставлении
одинаковых частей годового цикла. В B, шаг I.11, нулевая гипотеза состоит
в отсутствии систематического межгодового дрейфа после учёта сезона
и гидрологического состояния. Сопоставление одинаковых месяцев — предлагаемая
здесь ограниченная адаптация; оно не заменяет сопоставления влажности и поливов.

Пусть $D_{y,m}$ — фактически допустимые даты месяца $m$ года $y$ из основного
блока, $n_{y,m}=|D_{y,m}|$. Для амплитуды предлагается начать с медианы:

$$
\widetilde A_{j,y,m}=\mathrm{median}_{d\in D_{y,m}} A_j(d),
\qquad
\Delta A_{j,m}^{y_2-y_1}=\widetilde A_{j,y_2,m}-\widetilde A_{j,y_1,m}.
$$

Эта формула — рабочая адаптация для главы 3, а не алгоритм, дословно заданный
в PDF. Аналогичное сопоставление времён максимумов использует круговые сводки
следующего пункта. Нулевая гипотеза для выбранной сводки — отсутствие
межгодового сдвига; её проверка требует оценки неопределённости.

**Что сделать.** Сначала составить таблицу доступных месяцев 2017–2020
с числом и расположением дат. Затем выбрать сравнимые окна, например
октябри 2017/2018/2019, только если покрытие это допускает. Сравнить все три
уровня, сохранив пропуски и не присоединяя сентябрь 2013 к основному блоку.
Неравное число суток и наблюдения только в начале или конце месяца должны
быть видны рядом с результатом. Способ учёта неравного покрытия ещё предстоит выбрать.

**Будущий результат.** Таблица межгодовых различий в ppm и часах, использованных
дат, неопределённости и ограничений. Повторяемое календарное различие само
по себе не доказывает необратимой эволюции почвы: гидрологические условия
и приборный дрейф могут оставаться неучтёнными.


**Источник:** [B, §5, шаг I.11, стр. 15; §29, пункт 13, стр. 47](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=15).

[P, §19.1–19.2, стр. 25: сопоставимые состояния и отделение медленного изменения от внешних воздействий](../Papers/phase_topology_CO2_LEO.pdf#page=25).

#### 3.5.5.2. Круговая статистика времён максимумов

**Суть.** Время максимума лежит на окружности длиной 24 ч. Сравнивать его
среднее и разброс необходимо с учётом отождествления полуночи: 23:50 и 00:10
сосредоточены около 00:00. P, формула (66), задаёт взвешенное круговое среднее.

Адаптация к временам максимумов главы 2:

$$
\theta_j(d)=\frac{2\pi h_j(d)}{24\,\mathrm{ч}},\qquad
z_j(W)=\frac{\sum_{d\in W}w_d e^{i\theta_j(d)}}{\sum_{d\in W}w_d},
\quad w_d\geq0,\quad \sum_{d\in W}w_d>0,
$$

$$
\bar h_j(W)=\frac{24\,\mathrm{ч}}{2\pi}\arg z_j(W)
\pmod{24\,\mathrm{ч}},\qquad R_j(W)=|z_j(W)|.
$$

$\theta_j$ здесь — круговая координата времени максимума; знак фазового
коэффициента исходной гармоники этим не переопределяется. В качестве исходного
кандидата можно рассмотреть равные веса $w_d=1$. Амплитудное взвешивание
изменяет смысл сводки и требует отдельного обоснования.

Для сравнения двух окон предлагается кратчайшая разность времени:

$$
\Delta h_j=\mathrm{wrap}_{24}\bigl(\bar h_j(W_2)-\bar h_j(W_1)\bigr),
\qquad
\mathrm{wrap}_{24}(u)=((u+12\,\mathrm{ч})\bmod24\,\mathrm{ч})-12\,\mathrm{ч}.
$$

**Что сделать.** Вместе со средним показывать $R_j$, распределение времён
и круговой интервал неопределённости. При $R_j=0$ среднее не определено;
при малом $R_j$ оно неустойчиво, в том числе при нескольких группах пиков.
Численный порог здесь не назначается. Не разворачивать фазу через разрывы
или участки, где её ветвь не определяется. Разность по кратчайшей дуге
не определяет число полных оборотов за большой временной разрыв.

**Будущий результат.** Круговые сводки и межгодовые сдвиги в часах с явной
отметкой окон, где одна средняя фаза не представляет данные надёжно.


**Источник:** [P, §20.2, стр. 26, формула (66); §4.3, стр. 7](../Papers/phase_topology_CO2_LEO.pdf#page=26).

#### 3.5.5.3. Блочный bootstrap для неопределённости межгодовых различий

**Суть.** Соседние сутки могут быть зависимы. P, §20.2, рекомендует временной
блочный bootstrap: повторные выборки сохраняют последовательные участки
наблюдений вместо независимого перемешивания отдельных дат.

Пусть $T$ — заранее выбранная статистика сравнения, например $\Delta A_j$
или круговая $\Delta h_j$. Для $B_{\mathrm{rep}}$ повторных выборок:

$$
T^{*(b)}=T\bigl(W_1^{*(b)},W_2^{*(b)}\bigr),
\qquad b=1,\ldots,B_{\mathrm{rep}}.
$$

Для линейной разности амплитуд кандидат на процентильный интервал уровня
$1-\alpha$ имеет вид

$$
I_A=\bigl[Q_{\alpha/2}(\Delta A^*),\ Q_{1-\alpha/2}(\Delta A^*)\bigr].
$$

Формулы выше конкретизируют применение bootstrap к нашей задаче; PDF
не задаёт именно этот интервал. Для времени максимума рассматриваются
круговые отклонения $e_h^{*(b)}=\mathrm{wrap}_{24}(\Delta h^{*(b)}-\Delta h)$.
Линейные квантили этих отклонений допустимы только при однозначной локальной
ветви и достаточной концентрации; иначе нужен круговой доверительный набор
или вывод о неопределённости сдвига.

**Что сделать.** До реализации выбрать длину временного блока $\ell$,
число повторов, уровень интервала и способ обработки неполного календаря.
Блоки не должны пересекать исходные разрывы или смешивать несопоставимые
сезоны. При совместном анализе трёх уровней выбирать для них одни и те же
даты, сохраняя зависимость между датчиками. Проверить чувствительность к $\ell$.
Если непрерывных фрагментов недостаточно, указать ограничение применимости.

**Будущий результат.** Интервалы неопределённости выбранных различий
и диагностика их устойчивости. Bootstrap внутри имеющихся окон не заменяет
независимого воспроизведения на другом доступном межгодовом сравнении.


**Источник:** [P, §20.1–20.2, стр. 26; B, §29, пункт 6, стр. 47](../Papers/phase_topology_CO2_LEO.pdf#page=26).

[B: проверка на временных окнах](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=47).

#### 3.5.5.4. Устойчивость к выбору окна и календарному покрытию

**Суть.** Изменение должно сохранять научный смысл при разумных вариантах
сопоставления. В B, §29, предусмотрены разные ширины окон и разделение
надёжных и сомнительных периодов; P, §5.2, связывает полезность фаз
с устойчивостью к выбору процедуры.

Для заранее определённого набора вариантов $\mathcal K$ можно сохранять
разности $\Delta A_j^{(k)}$, $\Delta h_j^{(k)}$. Предлагаемые описательные
меры чувствительности относительно основного варианта $k_0$:

$$
S_{A,j}=\max_{k\in\mathcal K}\Delta A_j^{(k)}-
\min_{k\in\mathcal K}\Delta A_j^{(k)},
$$

$$
S_{h,j}=\max_{k\in\mathcal K}
\left|\mathrm{wrap}_{24}\bigl(\Delta h_j^{(k)}-\Delta h_j^{(k_0)}\bigr)\right|.
$$

Это предложенные для главы 3 сводки, а не пороги из источников.
Если для какого-либо варианта средняя фаза не определена, это отдельный
результат чувствительности, а не нулевой вклад в $S_{h,j}$.

**Что сделать.** Зафиксировать варианты границ и ширины окон сравнения
готовых суточных оценок, способ сопоставления календарного покрытия и длины
блоков bootstrap. Сохранять причины изменения набора дат. Не подбирать
варианты после просмотра результата, не повышать статус 1244 предварительных
точек и трёх пограничных дат. Новая подгонка суточной гармоники и новые
автоматические пороги качества этим пунктом не вводятся.

**Будущий результат.** Таблица «вариант — даты — разность — неопределённость —
ограничение». Для каждого уровня отдельно установить, насколько вывод
зависит от процедуры. Положительный результат по амплитуде не подтверждает
автоматически изменение времени максимума. При множестве сравнений заранее
определить основные утверждения и способ учёта множественности.


**Источник:** [B, §29, пункты 5 и 8, стр. 47](../Papers/LEO_Buchstaber_Deymier_project_soil_transformation_clarified.pdf#page=47).

[P, §5.2, стр. 8; §20.1, пункт 8, стр. 26](../Papers/phase_topology_CO2_LEO.pdf#page=8).

## 3.6. Форма будущего представления результата

После выполнения операций 3.5.1–3.5.4 по выбранной процедуре зафиксировать проверяемое сопоставление длительных изменений **всех трёх компонент** времён максимумов и амплитуд. Результат должен содержать использованные даты и статусы, точное описание применённой процедуры, выявленные изменения либо обоснование их неустановленности, ограничения чувствительности к пропускам и надёжности фаз.

| Предмет вывода | Результат | Основание и ограничения |
|---|---|---|
| Длительные изменения $\mathbf H_v(d)$ на основном блоке | Не установлено — анализ не выполнен | Заполнить после исследования |
| Длительные изменения $\mathbf A_v(d)$ на основном блоке | Не установлено — анализ не выполнен | Заполнить после исследования |
| Достаточность данных и воспроизводимость результата | Не установлено — анализ не выполнен | Заполнить после исследования |

Здесь «не установлено — анализ не выполнен» обозначает **отсутствие результата**, а не отрицательное научное заключение. Независимое рассмотрение трёх координат не требует вводить ещё одну модель либо автоматически анализировать междатчиковые относительные фазы.
